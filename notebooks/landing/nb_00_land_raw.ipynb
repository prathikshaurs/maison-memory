# nb_00_land_raw
Unzips the raw synthetic files from Files/maison_raw.zip into Files/landing/raw.  
Input: Files/maison_raw.zip  
Output: Files/landing/raw/<region>/dt=<date>/*.csv

```python
import zipfile, os

base = "/lakehouse/default/Files"
with zipfile.ZipFile(f"{base}/maison_raw.zip") as z:
    z.extractall(f"{base}/landing")

count = sum(len(files) for _, _, files in os.walk(f"{base}/landing/raw"))
print("files landed:", count)
```
